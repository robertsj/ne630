```python
import os
import openmc
import numpy as np
import matplotlib.pyplot as plt

xs_path = os.environ.get("OPENMC_CROSS_SECTIONS").replace("cross_sections.xml", "neutron/")

energy = np.logspace(-3, 7, 20000)
U238 = openmc.data.IncidentNeutron.from_hdf5(xs_path + "U238.h5")
u238_inelastic = U238[4].xs["294K"]
u238_fission   = U238[18].xs["294K"]

purple = "#512888"
orange = "#CA7C1B"
fig, ax = plt.subplots(figsize=(8, 3))
ax.loglog(energy, u238_inelastic(energy), color=purple, lw=1, label=r"$(n,n')$")
ax.loglog(energy, u238_fission(energy), color=orange, lw=1, label=r"$(n,f)$")
ax.set_xlabel("Energy [eV]")
ax.set_ylabel(r"$\sigma$ [b]")
ax.grid(True, which="both", alpha=0.25)
ax.legend(frameon=False)
plt.tight_layout()
plt.savefig("u238_threshold.pdf")
```
